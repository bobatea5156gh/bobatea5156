## Question 7: Computing Derivative Using PyTorch Autograd

Consider the function

$$
\tau = 2r \sqrt{\frac{r}{r_s}} \sqrt{1-\frac{r_s}{r}}
$$

Compute the derivative $\frac{d\tau}{dr}$ using PyTorch Autograd, given $r=30000.1$ and $r_s=0.00887$.



```python

import torch

# Given values
r_s = 0.00887
r = torch.tensor(30000.1, dtype=torch.float64, requires_grad=True)

# Calculate tau
tau = 2 * r * torch.sqrt(r / r_s) * torch.sqrt(1 - r_s / r)

# Compute derivative using PyTorch Autograd
tau.backward()

# Display results
print("tau =", tau.item())
print("d(tau)/dr =", r.grad.item())
```


**Result:**

Using PyTorch Autograd, the derivative of $\tau$ with respect to $r$ at $r=30000.1$ and $r_s=0.00887$ is approximately

$$
\frac{d\tau}{dr}=5517.2260
$$



## Question 8: Linear Regression in One Variable with Scikit-Learn

We use steer angle as the independent variable and yaw rate as the response variable. The dataset is split into 75% training data and 25% test data.


```python

import pandas as pd

# Load the RAV4 CAN messages dataset
url = "https://raw.githubusercontent.com/rahulbhadani/Datasets/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv"

df = pd.read_csv(url)

# Display the dataset size and column names
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

# Display the first five rows
df.head()
```

```python
# Display all column names clearly
for col in df.columns:
    print(col)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Select independent and response variables
X = df[["steer_angle"]]
y = df["yaw_rate"]

# Split into 75% training and 25% testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Train the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict using test data
y_pred = model.predict(X_test)

# Report model coefficients and test R-squared
print("Coefficient (slope):", model.coef_[0])
print("Intercept:", model.intercept_)
print("Test R-squared:", r2_score(y_test, y_pred))
```


### Results

The linear regression model was trained using 75% of the dataset and tested using the remaining 25%.

The estimated coefficient (slope) is **0.0662085**, and the intercept is **-0.2071971**.

Therefore, the fitted linear regression model is

$$
\widehat{\text{yaw rate}}
= -0.2071971 + 0.0662085(\text{steer angle})
$$

The coefficient of determination on the test data is

$$
R^2 = 0.6555644
$$

This means that the model explains approximately **65.56% of the variation** in yaw rate in the test dataset.
