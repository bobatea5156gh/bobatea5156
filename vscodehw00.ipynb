```python
from bobatea5156.differential.discrete import diff
```

```python
t = [0, 0.1, 0.3, 0.4, 0.55, 0.67, 0.71]
x = [23.1, 22.5, 23.5, 21.88, 22.5, 23.5, 24.88]

v = diff(t, x)

print("t =", t)
print("x =", x)
print("v =", v)
```

```python
import matplotlib.pyplot as plt

# Plot 1: x vs t
plt.figure()
plt.plot(t, x, marker='o')
plt.xlabel('t')
plt.ylabel('x')
plt.title('Signal x vs Time t')
plt.grid(True)
plt.savefig('x_vs_t.pdf')
plt.show()

# Plot 2: v vs t
plt.figure()
plt.plot(t[1:], v, marker='o')
plt.xlabel('t')
plt.ylabel('v')
plt.title('Discrete Derivative v vs Time t')
plt.grid(True)
plt.savefig('v_vs_t.pdf')
plt.show()
```
