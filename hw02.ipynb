## Part 1 — Disjoint Sets

Given:

$$
P(A)=\frac{1}{3}
$$

and

$$
P(B^c)=\frac{1}{4}.
$$

First, find $P(B)$:

$$
P(B)=1-P(B^c)
$$

$$
P(B)=1-\frac{1}{4}=\frac{3}{4}.
$$

If $A$ and $B$ are disjoint, then:

$$
P(A\cup B)=P(A)+P(B).
$$

Therefore:

$$
P(A\cup B)
=
\frac{1}{3}+\frac{3}{4}
=
\frac{4}{12}+\frac{9}{12}
=
\frac{13}{12}.
$$

But a probability cannot be greater than 1, and:

$$
\frac{13}{12}>1.
$$

Therefore, **A and B cannot be disjoint.**

## Part 2 — Moment Generating Function (MGF)

For a continuous random variable, the moment generating function is

$$
M_X(t)=E(e^{tX})
$$

or equivalently,

$$
M_X(t)=\int_{-\infty}^{\infty}e^{tx}f_X(x)\,dx.
$$

For a standard normal random variable,

$$
X\sim N(0,1),
$$

and its probability density function is

$$
f_X(x)=\frac{1}{\sqrt{2\pi}}e^{-x^2/2}.
$$

Therefore,

$$
M_X(t)
=
\int_{-\infty}^{\infty}
e^{tx}
\frac{1}{\sqrt{2\pi}}e^{-x^2/2}\,dx.
$$

Combine the exponential terms:

$$
M_X(t)
=
\frac{1}{\sqrt{2\pi}}
\int_{-\infty}^{\infty}
e^{tx-x^2/2}\,dx.
$$

Complete the square:

$$
tx-\frac{x^2}{2}
=
-\frac{1}{2}(x-t)^2+\frac{t^2}{2}.
$$

Thus,

$$
M_X(t)
=
e^{t^2/2}
\frac{1}{\sqrt{2\pi}}
\int_{-\infty}^{\infty}
e^{-(x-t)^2/2}\,dx.
$$

The integral is equal to $\sqrt{2\pi}$, so

$$
\boxed{M_X(t)=e^{t^2/2}}.
$$

Therefore, the MGF of the standard normal distribution is

$$
\boxed{M_X(t)=e^{t^2/2}}.
$$

## Part 3 — Probability Density Function

The given function is

$$
f(x)=\frac{4}{\beta^3\sqrt{\pi}}
x^2e^{-x^2}e^{-x^2/\beta^2},
\qquad 0<x<\infty,\quad \beta>0.
$$

To be a valid probability density function, two conditions must be satisfied:

1. $f(x)\geq 0$
2. $\displaystyle \int_{-\infty}^{\infty}f(x)\,dx=1$

### Condition 1: Non-negativity

Since $\beta>0$, $x^2\geq0$, and exponential functions are always positive,

$$
f(x)\geq0.
$$

Therefore, the first condition is satisfied.

### Condition 2: Total probability

Because $f(x)=0$ outside $x>0$, we calculate

$$
\int_{-\infty}^{\infty}f(x)\,dx
=
\int_0^\infty
\frac{4}{\beta^3\sqrt{\pi}}
x^2e^{-x^2}e^{-x^2/\beta^2}\,dx.
$$

Combine the exponential terms:

$$
e^{-x^2}e^{-x^2/\beta^2}
=
e^{-(1+1/\beta^2)x^2}.
$$

Therefore,

$$
\int_0^\infty f(x)\,dx
=
\frac{4}{\beta^3\sqrt{\pi}}
\int_0^\infty
x^2e^{-(1+1/\beta^2)x^2}\,dx.
$$

Using

$$
\int_0^\infty x^2e^{-ax^2}\,dx
=
\frac{\sqrt{\pi}}{4a^{3/2}},
\qquad a>0,
$$

with

$$
a=1+\frac{1}{\beta^2},
$$

we obtain

$$
\int_0^\infty f(x)\,dx
=
\frac{1}{\beta^3
\left(1+\frac{1}{\beta^2}\right)^{3/2}}.
$$

Since $\beta>0$,

$$
\beta^3
\left(1+\frac{1}{\beta^2}\right)^{3/2}
=
(\beta^2+1)^{3/2}.
$$

Thus,

$$
\int_{-\infty}^{\infty}f(x)\,dx
=
\frac{1}{(\beta^2+1)^{3/2}}.
$$

For $\beta>0$,

$$
\frac{1}{(\beta^2+1)^{3/2}}<1.
$$

Therefore, the total probability is not equal to 1.

### Conclusion

The function satisfies the non-negativity condition, but its integral is not equal to 1.

$$
\boxed{\text{Therefore, } f(x)\text{ is not a valid probability density function.}}
$$

## Part 6 — Probability Density Functions: Normal Distribution

Generating 10000 random samples from a Normal distribution with a fixed mean
$\mu=0.5$ and standard deviations $\sigma=0.3$, $5.0$, and $10.0$.

For a Normal distribution,

$$
X\sim N(\mu,\sigma^2)
$$

where $\mu$ determines the center of the distribution and $\sigma$ determines its spread.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

mu = 0.5
sigmas = [0.3, 5.0, 10.0]
n = 10000

fig, ax = plt.subplots(3, 1, figsize=(8, 12))

for i, sigma in enumerate(sigmas):
    samples = np.random.normal(mu, sigma, n)

    sns.histplot(
        samples,
        ax=ax[i],
        stat="density",
        kde=True
    )

    ax[i].set_title(f"Normal Distribution: μ = {mu}, σ = {sigma}")
    ax[i].set_xlabel("x")
    ax[i].set_ylabel("Density")

plt.tight_layout()
plt.show()
```

### Observation

The mean $\mu=0.5$ is kept fixed, so all three distributions are centered around 0.5.

When $\sigma=0.3$, the distribution is narrow and concentrated around the mean.

When $\sigma=5.0$, the distribution becomes much wider and flatter.

When $\sigma=10.0$, the distribution becomes even wider and flatter.

Therefore, increasing the standard deviation increases the spread of the probability density function while keeping its center approximately at $\mu=0.5$.

## Part 7 — Data Exploration with Motorway Control System (MCS)

### 7.1 Read the MCS Dataset

```python
import polars as pl

df = pl.read_csv("mcsOct2018.csv")

print(df)
print("Number of columns:", len(df.columns))
```

### 7.2 Convert Date to POSIX Time and Sort

Converting the `date` column to POSIX time and sort the dataset chronologically by date.

```python
# Convert date to POSIX time
df = df.with_columns(
    pl.col("date").str.strptime(
        pl.Datetime,
        format="%Y-%m-%d %H:%M:%S"
    ).dt.epoch(time_unit="s").alias("date")
)

# Sort chronologically
df = df.sort("date")

print(df.head())
```

### 7.3 Records per Sensor

Counting the number of records contributed by each `fk_id` and calculating the percentage of the total dataset contributed by each sensor.

```python
total_records = df.height

records_per_sensor = (
    df.group_by("fk_id")
    .agg(
        pl.len().alias("record_count")
    )
    .with_columns(
        (pl.col("record_count") / total_records * 100)
        .alias("percentage")
    )
    .sort("fk_id")
)

print(records_per_sensor)
```

### 7.4 Number of Lanes Used

Replacing NULL lane values with `np.nan`, count the valid lanes for each entry, and adding a new column called `num_lanes_used`.

```python
import numpy as np

# Replace NULL with NaN
df = df.with_columns(
    pl.col("used_lanes")
    .str.replace_all("NULL", "nan")
    .alias("used_lanes")
)

# Count only valid (non-NaN) lanes
df = df.with_columns(
    pl.col("used_lanes")
    .str.strip_chars("{}")
    .str.split(",")
    .list.eval(
        pl.element()
        .str.strip_chars()
        .cast(pl.Float64, strict=False)
    )
    .list.eval(
        pl.element().is_not_nan()
    )
    .list.sum()
    .alias("num_lanes_used")
)

print(
    df.select(
        ["fk_id", "date", "used_lanes", "num_lanes_used"]
    ).head(10)
)
```

### 7.5 Identify Potential Sensor Anomalies

```python
df = df.with_columns(
    (
        (pl.col("speed_std_dev") > 10) |
        (pl.col("flow") == 0)
    ).alias("is_anomaly")
)

print(
    df.select(
        ["fk_id", "date", "speed_std_dev", "flow", "is_anomaly"]
    )
    .filter(pl.col("is_anomaly"))
    .head(10)
)
```

## 9.3 Poisson Distribution — Inverse Transform Sampling

For a Poisson random variable with parameter $\lambda$, the probability mass function is

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!},
\qquad k=0,1,2,\ldots
$$

The cumulative distribution function is

$$
F(k)=P(X\leq k)
=\sum_{i=0}^{k}\frac{e^{-\lambda}\lambda^i}{i!}.
$$

For inverse transform sampling, generate a uniform random number

$$
U\sim Uniform(0,1).
$$

Then find the smallest integer $k$ such that

$$
F(k)\geq U.
$$

That value of $k$ is the generated Poisson random sample.

```python
from difftest.distributions.cvdistributions import poissondist

print(poissondist(1.0))
print(poissondist(4.0))
print(poissondist(10.0))
```

```python
import matplotlib.pyplot as plt

samples = [poissondist(4.0) for _ in range(100000)]

plt.hist(samples, bins=range(min(samples), max(samples) + 2), edgecolor="black")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of 100,000 Poisson Samples (λ = 4)")
plt.show()
```
