```python
from bobatea5156.differential.discrete import diff
```

```python
t = [0, 0.1, 0.3, 0.4, 0.55, 0.67, 0.71]

x = [23.1, 22.5, 23.5, 21.88, 22.5, 23.5, 24.88]
```

```python
v = diff(x, t)
print(v)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

# (i) x vs t
plt.subplot(2, 1, 1)
plt.plot(t, x, marker='o')
plt.xlabel('t')
plt.ylabel('x')
plt.title('x vs t')
plt.grid(True)

# (ii) v vs t
plt.subplot(2, 1, 2)
plt.plot(t[1:], v, marker='o')
plt.xlabel('t')
plt.ylabel('v')
plt.title('Discrete derivative v vs t')
plt.grid(True)

plt.tight_layout()

# Save both plots as one PDF
plt.savefig('plots.pdf')
plt.show()
```
